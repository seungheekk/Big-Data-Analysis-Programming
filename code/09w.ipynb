```python
import matplotlib.pyplot as plt
#데이터 준비
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [2, 4, 6, 8]
y3 = [3, 6, 9,12]
y4 = [4, 8, 12, 16]
#선 스타일별로 그래프 그리기
plt.plot(x, y1, linestyle='-', label='Solid line')  #실선
plt.plot(x, y2, linestyle='--', label='Dashed line')  #점
plt.plot(x, y3, linestyle=':', label='Dotted line')  #도트 점선
plt.plot(x, y4, linestyle='-.', label='Dash-dot line')  #대시 도트 점선

#그래프 제목과 축 레이블
plt.title("선 종류(Line Styles in Matplotlib)")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

#범례 추가
plt.legend()
#그래프 보여주기
plt.show()
```

```python
import matplotlib.pyplot as plt

#데이터 준비
x = [1, 2, 3,4]
y1= [1, 4, 9, 16]
y2 = [2, 3, 4, 5]
y3 = [2, 3, 4, 5]
y4 = [3, 1, 4,2]
y5 = [4, 5, 6,8]

#그래프 그리기
plt.plot(x, y1, marker='o', label='Circle')
plt.plot(x, y2, marker='^',label='Triangie_up') #삼각형 위
plt.plot(x, y3, marker='s', label='Square') #사각형
plt.plot(x, y4, marker='*', label='Strar') #별
plt.plot(x, y5, marker=',', label='Pixel') # 픽셀
#그래프 제목과 축 레이블
plt.title("Marker Styles in Matplotlib")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
#범례 추가
plt.legend()
#그래프 보여주기
plt.show()
```

##빈도 수 값을 막대 그래프에 표시

```python

import seaborn as sns
ax = sns.countplot(data = mtc, x='gear',
                    order = mtc.gear.value_counts().index)
for p in ax.patches:
  height = p.get_height() #빈도수
  ax.text(x = p.get_x() + p.get_width()/2, y = (height + .2),
               s = int(height), ha = 'center', size = 12, color = 'blue')

ax.set_ylim(0,17) #y출의 범위를 0부터17까지로 설정
plt.show()

```
